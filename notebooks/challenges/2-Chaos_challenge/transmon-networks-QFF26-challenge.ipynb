## 🔬 The Physics: From Capacitors to Qubits

Superconducting transmon circuits can be modeled by a capacitance matrix $C$. The diagonal elements represent the self-capacitance of each qubit, and the off-diagonal elements represent the couplings between them.

When we invert this matrix to find the charging energies, we generate the interaction strengths $J_{ij}$ between the qubits. The effective Hamiltonian of this transmon network can be mapped to an interacting spin-1/2 model:

$$ H = \sum_{i<j} J_{ij} (X_i X_j + Y_i Y_j) + \sum_{i} \omega_i Z_i $$

Where:
* $X, Y, Z$ are the Pauli matrices.
* $J_{ij} \propto (C^{-1})_{ij}$ is the coupling strength between qubit $i$ and qubit $j$.
* $\omega_i$ is the effective frequency of qubit $i$.

### 1.2 From Capacitance to the Transmon Hamiltonian

To map our classical circuit to a quantum spin model, we must derive the charging energy matrix $E_C$. This is found simply by inverting the capacitance matrix: $E_C = C^{-1}$.

*(Note: For this hackathon simulation, we use normalized units, absorbing the electron charge constants, though the formal expression is $E_C = e^2C^{-1}/2$).*

Once we compute $E_C$, we extract the two fundamental physical parameters governing the transmons:

**1. Effective Qubit Frequencies ($\omega_i$)**  
The diagonal elements of the charging energy matrix, combined with the Josephson energy $E_J$, determine the resonance frequency of each transmon. This acts as the coefficient for the local $Z$ terms. In the highly anharmonic transmon regime, this is approximated as:
$$ \omega_i = \sqrt{8 E_J (E_C)_{ii}} - (E_C)_{ii} $$

**2. Exchange Couplings ($J_{ij}$)**  
The off-diagonal elements of the inverse capacitance matrix dictate the interaction strength between qubits $i$ and $j$. This manifests as an exchange interaction (photon hopping) in the XY plane:
$$ J_{ij} = (E_C)_{ij} $$

Putting it all together, the effective Hamiltonian in the Pauli basis is:
$$ H = \sum_{i<j} J_{ij} (X_i X_j + Y_i Y_j) + \sum_{i} \omega_i Z_i $$

## Part 2: The Signatures of Chaos (Exact Diagonalization)

How can we tell if a quantum system is integrable or chaotic? We look at its energy spectrum.

In an **integrable system**, energy levels are uncorrelated, they don't "care" about each other and can cluster together closely. This results in a **Poisson distribution** of energy gaps.
In a **chaotic system**, energy levels exhibit **level repulsion**, they push away from each other, resulting in a **Wigner-Dyson distribution** (characteristic of Random Matrix Theory).

To quantify this, we calculate the **mean level spacing ratio**, $\langle r \rangle$.

Here is the step-by-step math:
1. **Find the Energies:** Compute the sorted eigenvalues (energy levels) $E_n$ of the Hamiltonian.
2. **Find the Gaps:** Calculate the distance between adjacent energy levels: $\delta_n = E_{n+1} - E_n$.
3. **Filter Degeneracies:** Quantum systems often have symmetries that cause exact degeneracies ($\delta_n \approx 0$). These will artificially skew our statistics, so we must filter out any gaps that are effectively zero.
4. **Compute the Ratio:** For adjacent gaps, calculate the ratio of the smaller gap to the larger gap:
$$ r_n = \frac{\min(\delta_n, \delta_{n+1})}{\max(\delta_n, \delta_{n+1})} $$
5. **Take the Mean:** Average all $r_n$ values to find $\langle r \rangle$.

### 📊 The Reference Numbers
* **$\langle r \rangle \approx 0.386$** $\rightarrow$ Integrable (Poisson)
* **$\langle r \rangle \approx 0.530$** $\rightarrow$ Chaotic (Wigner-Dyson / GOE)

 ## Part 3: The Spectral Form Factor (Dynamical Chaos)

Energy level statistics tell us if a system is chaotic, but they require us to find all the eigenvalues, a task that becomes impossible as the quantum processor grows. To measure chaos in experimental hardware, we need a **dynamical** signature that evolves over time.

The gold standard for this is the **Spectral Form Factor (SFF)**, denoted as $K(t)$.

Mathematically, it is the partition function analytically continued to imaginary time. In the context of your Hamiltonian's eigenenergies $E_n$, it is defined as:

$$ K(t) = \frac{1}{D^2} \left\langle \sum_{n,m=1}^{D} e^{-i (E_n-E_m) t} \right\rangle = \frac{1}{D^2} \left\langle \left| \sum_{n=1}^{D} e^{-i E_n t} \right|^2 \right\rangle $$

Where:
*   $t$ is the evolution time.
*   $D$ is the dimension of the Hilbert space (the total number of energy levels).
*   $\langle \dots \rangle$ denotes an **ensemble average** over many slightly different disorder realizations (simulating the natural microscopic variations in real superconducting circuits).

### The Signature of Hardware Scrambling: The Dip and Damping
In an idealized, fully chaotic theoretical model (like Gaussian Orthogonal Ensembles), $K(t)$ on a log-log scale shows a perfectly distinct "dip, ramp, and plateau." However, physical finite-size hardware models—like our native 1D transmon chain—reveal a more complex, oscillatory reality governed by the strength of their internal coupling:

1.  **The Dip:** An initial sharp decay as the state dephases, plunging to a deep minimum.
2.  **The Oscillatory Recovery:** Instead of a smooth, steady upward ramp, finite 1D transmon arrays exhibit persistent, wild oscillations. This occurs because the finite system's energy levels maintain strong commensurability.
3.  **Coupling-Induced Damping:** This is where the physics of information scrambling appears! As you increase the non-local coupling ($C_{nn}$), level repulsion emerges. This spectral rigidity prevents the energy levels from perfectly destructively interfering, which causes the initial dip to become much **shallower**, and actively **damps** the wild oscillations in the intermediate time regime.
4.  **The Plateau:** The oscillations eventually die out and saturate at $K(t) = 1/D$, a limit dictated by the finite size of the Hilbert space.

### 3.1 Measuring the SFF with Quantum Circuits (Detour)

So far, every number in this notebook came from diagonalizing $H$ on a classical computer, which is exactly what a real experiment cannot do. A quantum processor can only **prepare** states, **evolve** them and **measure** them. Remarkably, that is enough to measure $K(t)$!

**The trick: a randomized echo.** For a random state $|\psi\rangle$ drawn uniformly (Haar measure) from the $D$-dimensional Hilbert space, the average *survival probability* is directly related to the SFF:

$$ \overline{P}(t) = \mathbb{E}_{\psi}\left|\langle \psi | e^{-iHt} | \psi \rangle\right|^2 = \frac{D\,K(t) + 1}{D + 1} \quad\Longrightarrow\quad K(t) = \frac{(D+1)\,\overline{P}(t) - 1}{D}. $$

**The circuit.** $|\langle \psi | U(t) | \psi \rangle|^2$ is the probability of measuring $|0\dots0\rangle$ after: preparing $|\psi\rangle$ from $|0\dots0\rangle$ (Qiskit's `StatePreparation`), applying $U(t)$, and undoing the preparation.

**The time evolution.** On hardware, $U(t) = e^{-iHt}$ must be built from gates. We use the **Lie-Trotter** product formula, which splits $H = \sum_k h_k P_k$ into its Pauli terms and applies each one for a short time $\Delta t$:

$$ e^{-iHt} \approx \Big( \prod_k e^{-i h_k P_k \Delta t} \Big)^{t/\Delta t}, $$

available in Qiskit as `PauliEvolutionGate(H, time=t, synthesis=LieTrotter(reps=t/Δt))`. Here this works well with fairly large steps: the large $\omega_i Z_i$ terms are almost uniform, and $\sum_i Z_i$ commutes with the flip-flop terms $X_iX_j + Y_iY_j$ (it counts the conserved excitations). Only the small differences between the $\omega_i$ create Trotter error.

> **⚠️ Qiskit gotcha:** `Statevector(circuit)` and `Operator(circuit)` use the *exact* matrix exponential of a `PauliEvolutionGate`. To simulate the Trotter circuit that a device would actually run, `transpile` it to basis gates first.

# Part 4: Scaling to 2D Architectures

So far, we have studied quantum chaos in a **1D transmon chain**. Real superconducting processors, however, are typically arranged in two-dimensional layouts, where each qubit can interact with several surrounding qubits.

Moving from 1D to 2D changes the connectivity of the system significantly. In addition to **nearest-neighbor** interactions, we will now include **next-nearest-neighbor** couplings along the diagonals of the lattice.

The effective Hamiltonian takes the form

$$
H =
J_1\sum_{\langle i,j\rangle}(X_iX_j+Y_iY_j)
+
J_2\sum_{\langle\langle i,j\rangle\rangle}(X_iX_j+Y_iY_j)
+
\sum_i \omega_i Z_i ,
$$

where

* $\langle i,j\rangle$ denotes nearest-neighbor pairs,
* $\langle\langle i,j\rangle\rangle$ denotes next-nearest-neighbor pairs,
* $J_1$ and $J_2$ are obtained from the inverse capacitance matrix,
* and $\omega_i$ represents the effective local frequency of each transmon.

We will use **Open Boundary Conditions (OBC)**, so qubits located at opposite edges of the lattice are not artificially connected.

---

## The dimensionality problem

There is an important difficulty when scaling the system up.

For $N$ qubits, the dimension of the full Hilbert space is

$$
\dim(\mathcal H)=2^N.
$$

For a $5\times5$ lattice,

$$
N=25,
$$

and therefore

$$
2^{25}=33\,554\,432
$$

basis states.

Constructing and diagonalizing the full Hamiltonian is therefore no longer practical on a standard computer.

Fortunately, our Hamiltonian has an important symmetry.

The interaction

$$
X_iX_j+Y_iY_j
$$

can be written in terms of raising and lowering operators as a **flip-flop interaction**: an excitation can move from one qubit to another, but the total number of excitations is conserved.

Therefore,

$$
[H,N_{\mathrm{ex}}]=0.
$$

Instead of working in the complete Hilbert space, we can restrict the Hamiltonian to a sector with a fixed number of excitations.

For example, choosing

$$
N_{\mathrm{ex}}=3
$$

reduces the Hilbert-space dimension from

$$
2^{25}=33\,554\,432
$$

to only

$$
\binom{25}{3}=2300.
$$

This symmetry reduction is what makes the $5\times5$ problem computationally accessible.

---

## Building the reduced basis

A basis state in the three-excitation sector can be represented simply by the positions of the three excited qubits.

For example,

$$
(2,7,15)
$$

represents a state in which qubits 2, 7 and 15 are excited.

Python's

```python
from itertools import combinations
```

is useful for generating all possible states with exactly $N_{\mathrm{ex}}$ excitations:

```python
states = list(combinations(range(N), n_excitations))
```

We can then construct the Hamiltonian **directly inside this reduced basis**, without ever generating the full $2^N\times2^N$ matrix.

---

## Sparse Hamiltonian construction

Even after restricting the Hilbert space, most pairs of basis states are not directly connected.

For this reason, we will use sparse matrices from

```python
import scipy.sparse as sp
```

rather than storing every zero element of the Hamiltonian.

The diagonal part contains the local excitation energies,

$$
E_{\alpha}=\sum_{i\in\alpha}\omega_i,
$$

while the off-diagonal part represents an excitation hopping from an occupied qubit $i\in\alpha$ to an unoccupied qubit $j\notin\alpha$. Because the flip-flop interaction only moves one excitation at a time, two basis states $\vert{}\alpha\rangle$ and $\vert{}\beta\rangle$ are connected by a non-zero coupling ($2J_{ij}$) only when they differ by the position of a single excitation.

📝 **Note on Operator Conventions (Factors of $0.5$ and $2.0$)**:

When switching from Qiskit's full Pauli basis (Part 1) to the fixed-excitation number basis $\hat{n}_i \in \{0, 1\}$ (Part 4), you will notice two numerical prefactors in the Hamiltonian construction:
- Local Frequencies (0.5 * ...): In the Pauli basis, $Z_i = 2\hat{n}_i - I$ has eigenvalues $\pm 1$ (separated by $2$), whereas in the excitation basis we only sum over occupied sites ($\hat{n}_i = 1$). We include the standard circuit-QED factor of $1/2$ in $\omega_i = \frac{1}{2}\left(\sqrt{8(E_C)_{ii}E_{J,i}} - (E_C)_{ii}\right)$ and drop the constant identity offset $-\sum_i \omega_i I$, since a global energy shift cancels out identically when computing eigenvalue gaps $\delta_n = E_{n+1} - E_n$.
- Off-Diagonal Hopping (2.0 * J_ij): Expanding the Pauli exchange term into raising and lowering operators yields $J_{ij}(X_i X_j + Y_i Y_j) = 2 J_{ij}(\sigma_i^+ \sigma_j^- + \sigma_i^- \sigma_j^+)$, giving a matrix element of $2J_{ij}$ for each single-excitation hop.
